```python
class Dense:
    def __init__(self, weights, bias):
        self.weights = weights
        self.bias = bias

    def forward(self, x):
        total = self.bias
        for w, xi in zip(self.weights, x):
            total += w * xi
        return total

    def __repr__(self):
        return f"Dense(weights={self.weights}, bias={self.bias})"

layer = Dense([0.5, -1.0], 0.1)
print(layer)
print(layer.forward([2.0, 3.0]))
layer.bias = 0.0
print(layer.forward([2.0, 3.0]))
```

```python
import numpy as np

def relu(z):
    return np.maximum(0.0, z)

def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))

def forward(x, W1, b1, W2, b2):
    z1 = W1 @ x + b1          # (3,2)@(2,) + (3,)  -> (3,)
    a1 = relu(z1)             # elementwise        -> (3,)
    z2 = W2 @ a1 + b2         # (1,3)@(3,) + (1,)  -> (1,)
    y  = sigmoid(z2)          # output activation  -> (1,)
    return z1, a1, z2, y

x  = np.array([1.0, 2.0])
W1 = np.array([[0.5, -0.2], [0.3, 0.8], [-0.6, 0.1]])
b1 = np.array([0.1, -0.2, 0.05])
W2 = np.array([[0.7, -0.4, 0.9]])
b2 = np.array([-0.3])
z1, a1, z2, y = forward(x, W1, b1, W2, b2)
print("z1:", np.round(z1, 4))
print("a1:", np.round(a1, 4))
print("z2:", np.round(z2, 4), "shape", z2.shape)
print("y :", np.round(y, 4))
```

```python
import numpy as np
sig = lambda z: 1 / (1 + np.exp(-z))
x, y = np.array([1.0, 2.0]), 1.0                 # one example (x shape (2,)), target
def step(W1, b1, w2, b2, eta=0.5):
    z1 = W1 @ x + b1                              # (2,)  W1 shape (n_out=2, n_in=2)
    a1 = sig(z1)
    y_hat = w2 @ a1 + b2                          # scalar output (linear)
    dy = y_hat - y                                # dL/dy_hat for L = 0.5*(y_hat - y)^2
    dw2, da1 = dy * a1, dy * w2                   # gradients (2,)
    dz1 = da1 * a1 * (1 - a1)                     # chain rule through sigmoid
    dW1 = np.outer(dz1, x)                        # (2,2): one row per hidden neuron
    return (W1 - eta * dW1, b1 - eta * dz1, w2 - eta * dw2, b2 - eta * dy), y_hat
def train(W1, w2, n=3):
    p = (W1.copy(), np.zeros(2), w2.copy(), 0.0)
    for _ in range(n): p, yh = step(*p)
    return p
same = train(np.full((2, 2), 0.5), np.array([0.5, 0.5]))
print("equal init   -> W1 after 3 steps:\n", np.round(same[0], 6))
rng = np.random.default_rng(0)
rand = train(rng.normal(0, 0.5, (2, 2)), rng.normal(0, 0.5, 2))
print("random init  -> W1 after 3 steps:\n", np.round(rand[0], 6))
print("hidden rows identical? equal:", bool(np.allclose(same[0][0], same[0][1])), "| random:", bool(np.allclose(rand[0][0], rand[0][1])))
```

```python
import numpy as np
sigm = lambda z: 1 / (1 + np.exp(-z))
relu = lambda z: np.maximum(0, z)
def grad_norms(act, dact, std, L=10, n=64, seed=0):
    rng = np.random.default_rng(seed)
    Ws = [rng.normal(0, std, (n, n)) for _ in range(L)]   # W[l] shape (n_out, n_in)
    a, zs = rng.normal(0, 1, n), []
    for W in Ws:                                          # forward pass, keep z
        z = W @ a; zs.append(z); a = act(z)
    g, norms = np.ones(n), []                             # arbitrary upstream gradient dL/da[L] = 1
    for l in reversed(range(L)):
        g = g * dact(zs[l])                               # dL/dz[l]
        g = Ws[l].T @ g                                   # dL/da[l-1]
        norms.append(float(np.linalg.norm(g)))
    return norms[::-1]                                    # index 0 = earliest layer
d_sig = lambda z: sigm(z) * (1 - sigm(z)); d_relu = lambda z: (z > 0).astype(float)
cases = {"sigmoid, std 1/sqrt(n)": (sigm, d_sig, 1 / np.sqrt(64)), "ReLU, He": (relu, d_relu, np.sqrt(2 / 64)), "ReLU, std 0.35 (too big)": (relu, d_relu, 0.35)}

for k, (a_, d_, s_) in cases.items():
    r = grad_norms(a_, d_, s_)
    print(f"{k:<26} grad norm at layer 1: {r[0]:.3e} | layer 5: {r[4]:.3e} | layer 10: {r[9]:.3e}")
print("sigmoid'(z) at z = 0, 2, 5, 10:", np.round(d_sig(np.array([0., 2., 5., 10.])), 6))
```

```python
import numpy as np

def stratified_split(y, fractions=(0.7, 0.15, 0.15), seed=0):
    rng = np.random.default_rng(seed)
    out = {"train": [], "val": [], "test": []}
    for c in np.unique(y):                  # split each class
        ids = rng.permutation(np.where(y == c)[0])
        n_tr = int(round(fractions[0] * len(ids)))
        n_va = int(round(fractions[1] * len(ids)))
        out["train"] += ids[:n_tr].tolist()
        out["val"] += ids[n_tr:n_tr + n_va].tolist()
        out["test"] += ids[n_tr + n_va:].tolist()
    return {k: np.array(sorted(v)) for k, v in out.items()}
```

```python
import torch
import torch.nn as nn

class TinyClassifier(nn.Module):
    def __init__(self, in_features=2, hidden=16):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_features, hidden),   # (B,2)  -> (B,16)
            nn.ReLU(),
            nn.Linear(hidden, hidden),        # (B,16) -> (B,16)
            nn.ReLU(),
            nn.Linear(hidden, 1),             # (B,16) -> (B,1)
        )

    def forward(self, x):                     # x: (B, 2)
        return self.net(x).squeeze(1)         # logits: (B,)

model = TinyClassifier()
print(sum(p.numel() for p in model.parameters()))
print(model(torch.randn(8, 2)).shape)
```

```python
@torch.no_grad()                                 # no graph, no grads
def evaluate(model, loader, loss_fn, device):
    model.eval()                              # dropout off, BN frozen
    total, n, logit_list, y_list = 0.0, 0, [], []
    for xb, yb in loader:
        xb, yb = xb.to(device), yb.to(device)
        logits = model(xb)                       # (B,)
        total += loss_fn(logits, yb).item() * xb.size(0)
        n += xb.size(0)
        logit_list.append(logits.cpu())
        y_list.append(yb.cpu())
    return total / n, torch.cat(logit_list), torch.cat(y_list)

# no optimizer.step() and no backward() anywhere above
```

```python
import torch.nn as nn
class SmallCNN(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 16, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(16, 32, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2))
        self.head = nn.Sequential(nn.Flatten(), nn.Linear(32*8*8, num_classes))
    def forward(self, x): return self.head(self.features(x))  # [B,10]
```

```python
import torch, torch.nn as nn
device = "cuda" if torch.cuda.is_available() else "cpu"
model = SmallCNN().to(device); criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

def train_one_epoch(loader):
    model.train()                                     # enables dropout / BN training behaviour
    for x, y in loader:
        x, y = x.to(device), y.to(device)             # y: int64 [B]
        optimizer.zero_grad()
        loss = criterion(model(x), y)
        loss.backward(); optimizer.step()

@torch.no_grad()
def evaluate(loader):
    model.eval(); correct = total = 0
    for x, y in loader:
        pred = model(x.to(device)).argmax(1).cpu()    # argmax over class dim
        correct += (pred == y).sum().item(); total += len(y)
    return correct / total
```

```python
import numpy as np
rng = np.random.default_rng(0); yy, xx = np.mgrid[:32, :32]
def draw(cls):                                           # 0 circle, 1 square, 2 cross
    cy, cx, r = rng.integers(10, 22), rng.integers(10, 22), rng.integers(5, 9)
    if cls == 0:   m = (yy-cy)**2 + (xx-cx)**2 <= r*r
    elif cls == 1: m = (abs(yy-cy) <= r) & (abs(xx-cx) <= r)
    else:          m = ((abs(yy-cy) <= 1) & (abs(xx-cx) <= r+2)) | ((abs(xx-cx) <= 1) & (abs(yy-cy) <= r+2))
    return np.clip(m + rng.normal(0, 0.1, (32, 32)), 0, 1)
y = np.repeat([0, 1, 2], 300); rng.shuffle(y)
X = np.stack([draw(c) for c in y])[:, None].astype(np.float32)   # [N, 1, 32, 32]
tr, va, te = [], [], []
for c in range(3):                                       # stratified 70/15/15, done once
    ci = rng.permutation(np.where(y == c)[0]); tr += ci[:210].tolist(); va += ci[210:255].tolist(); te += ci[255:].tolist()
mu, sd = X[tr].mean(), X[tr].std()                       # statistics from TRAIN only
print("X", X.shape, X.dtype, "range", round(float(X.min()), 2), round(float(X.max()), 2))
print("split", len(tr), len(va), len(te), "| overlap", len(set(tr)&set(va) | set(tr)&set(te) | set(va)&set(te)))
print("train class counts", np.bincount(y[tr]), "| train mean/std", round(float(mu), 4), round(float(sd), 4))
```

```python
import numpy as np, torch, torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
torch.manual_seed(0)
device = "cuda" if torch.cuda.is_available() else "cpu"
def prep(idx): return torch.from_numpy((X[idx] - mu) / sd), torch.from_numpy(y[idx]).long()
(Xtr, ytr), (Xva, yva) = prep(tr), prep(va)            # X, y, tr, va, mu, sd come from Page 223
train_loader = DataLoader(TensorDataset(Xtr, ytr), batch_size=32, shuffle=True)
val_loader   = DataLoader(TensorDataset(Xva, yva), batch_size=64)
model = nn.Sequential(nn.Conv2d(1, 8, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
                      nn.Conv2d(8, 16, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
                      nn.Flatten(), nn.Linear(16*8*8, 3)).to(device)
opt = torch.optim.Adam(model.parameters(), lr=1e-3); crit = nn.CrossEntropyLoss(); best = 0.0
for epoch in range(15):
    model.train()
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        opt.zero_grad(); loss = crit(model(xb), yb); loss.backward(); opt.step()
    model.eval(); correct = 0
    with torch.no_grad():
        for xb, yb in val_loader: correct += (model(xb.to(device)).argmax(1).cpu() == yb).sum().item()
    acc = correct / len(yva)
    if acc > best: best = acc; torch.save(model.state_dict(), "best.pt")   # best on VALIDATION
    print(epoch, round(acc, 4))
```
